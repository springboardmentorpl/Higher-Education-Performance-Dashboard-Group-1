```python
df2["Overall Score"] = pd.to_numeric(
    df2["Overall Score"],
    errors="coerce"
)

print(df2["Overall Score"].dtype)
```

```python
import pandas as pd

# Dataset 1 - QS
df1 = pd.read_excel("/dataset(top).xlsx")

# Dataset 2 - THE
df2 = pd.read_csv("/dataset(qs).csv")

print("Dataset 1 shape:", df1.shape)
print("Dataset 2 shape:", df2.shape)
```

```python
print("Dataset 1 missing values:")
print(df1.isnull().sum())

print("\nDataset 2 missing values:")
print(df2.isnull().sum())
```

```python
missing_columns = [
    "International Faculty",
    "International Students",
    "International Research Network",
    "Sustainability"
]

for column in missing_columns:
    df2[column] = df2[column].fillna(df2[column].median())
```

```python
print("Dataset 1 missing values:")
print(df1.isnull().sum().sum())

print("\nDataset 2 missing values:")
print(df2.isnull().sum())
```

```python
# Clean column names
df1.columns = (
    df1.columns.astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

df2.columns = (
    df2.columns.astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

print("Dataset 1 columns:")
print(df1.columns.tolist())

print("\nDataset 2 columns:")
print(df2.columns.tolist())
```

```python
print("Dataset 1 university-related columns:")
print([col for col in df1.columns if "university" in col.lower()])

print("\nDataset 2 university-related columns:")
print([col for col in df2.columns if "university" in col.lower()])
```

```python
df1["University Name"] = (
    df1["University Name"]
    .astype(str)
    .str.strip()
)

df2["University Name"] = (
    df2["University Name"]
    .astype(str)
    .str.strip()
)
```

```python
common_universities = set(df1["University Name"]) & set(df2["University Name"])

print("Number of common universities:", len(common_universities))
```

```python
print("Dataset 1 rows:", len(df1))
print("Dataset 2 rows:", len(df2))

common_universities = set(df1["University Name"]) & set(df2["University Name"])

print("Common universities:", len(common_universities))
```

```python
merged_df = pd.merge(
    df1,
    df2,
    on="University Name",
    how="inner",
    suffixes=("_QS", "_THE")
)
```

```python
print("Merged dataset shape:", merged_df.shape)
```

```python
print("Number of universities in merged dataset:",
      merged_df["University Name"].nunique())
```

```python
display(merged_df.head())
```

```python
merged_df.to_excel(
    "/content/cleaned_merged_university_rankings.xlsx",
    index=False
)

print("File created successfully!")
```
