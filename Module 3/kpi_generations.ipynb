```python
import pandas as pd

df = pd.read_excel("/cleaned_merged_university_rankings.xlsx")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print(df.columns.tolist())
```

```python
# 1. Global Ranking Score

# Convert QS Rank into a score where higher rank score = better
qs_rank_score = (
    1 - (df["Rank"] - df["Rank"].min()) /
    (df["Rank"].max() - df["Rank"].min())
) * 100

# Convert THE 2025 Rank into a score where higher score = better
the_rank_score = (
    1 - (df["2025 Rank"] - df["2025 Rank"].min()) /
    (df["2025 Rank"].max() - df["2025 Rank"].min())
) * 100

# Calculate Global Ranking Score
df["Global Ranking Score"] = (
    qs_rank_score + the_rank_score
) / 2

# Round to 2 decimal places
df["Global Ranking Score"] = df["Global Ranking Score"].round(2)

# Display the result
display(
    df[
        ["University Name", "Rank", "2025 Rank", "Global Ranking Score"]
    ].head(10)
)
```

```python
# 2. Research Impact Score

df["Research Impact Score"] = (
    df["Research Quality"] +
    df["Citations per Faculty"]
) / 2

# Round to 2 decimal places
df["Research Impact Score"] = (
    df["Research Impact Score"].round(2)
)

# Display the result
display(
    df[
        [
            "University Name",
            "Research Quality",
            "Citations per Faculty",
            "Research Impact Score"
        ]
    ].head(10)
)
```

```python
# 3. Faculty-to-Student Ratio

df["Faculty-to-Student Ratio"] = (
    df["No. of Students per Staff"]
)

# Round to 2 decimal places
df["Faculty-to-Student Ratio"] = (
    df["Faculty-to-Student Ratio"].round(2)
)

# Display the result
display(
    df[
        [
            "University Name",
            "No. of Students per Staff",
            "Faculty-to-Student Ratio"
        ]
    ].head(10)
)
```

```python
# 4. International Student Percentage

df["International Student Percentage"] = (
    df["International Students_QS"] * 100
)

df["International Student Percentage"] = (
    df["International Student Percentage"].round(2)
)

display(
    df[
        [
            "University Name",
            "International Students_QS",
            "International Student Percentage"
        ]
    ].head(10)
)
```

```python
# 5. Academic Reputation Score

df["Academic Reputation Score"] = (
    df["Academic Reputation"]
)

df["Academic Reputation Score"] = (
    df["Academic Reputation Score"].round(2)
)

display(
    df[
        [
            "University Name",
            "Academic Reputation",
            "Academic Reputation Score"
        ]
    ].head(10)
)
```

```python
# 6. Research Productivity Index

df["Research Productivity Index"] = (
    df["Research Quality"] +
    df["International Research Network"]
) / 2

df["Research Productivity Index"] = (
    df["Research Productivity Index"].round(2)
)

display(
    df[
        [
            "University Name",
            "Research Quality",
            "International Research Network",
            "Research Productivity Index"
        ]
    ].head(10)
)
```

```python
# Check that all 6 KPI columns exist
kpi_columns = [
    "Global Ranking Score",
    "Research Impact Score",
    "Faculty-to-Student Ratio",
    "International Student Percentage",
    "Academic Reputation Score",
    "Research Productivity Index"
]

print("KPI columns found:")
print([col for col in kpi_columns if col in df.columns])
```

```python
# Save the cleaned dataset with the 6 KPI columns added

df.to_excel(
    "/content/university_final_dataset.xlsx",
    index=False
)

print("Final dataset saved successfully!")
```
