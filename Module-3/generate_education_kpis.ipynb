```python
df = df.drop_duplicates()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)
```

```python
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df = df.rename(columns={
    "total_students": "Total_Students",
    "intl_students_pct": "International_Students_Percentage"
})
```

```python
df["founded"] = pd.to_numeric(df["founded"], errors="coerce")
```

```python
# Data Transformation
```

```python
df["university_age"] = 2026 - df["founded"]
```

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(r"\s+", "_", regex=True)
```

```python
print(df.columns.tolist())
```

```python
df["Overall_Score_Normalized"] = (
    (df["Overall_SCORE"] - df["Overall_SCORE"].min()) /
    (df["Overall_SCORE"].max() - df["Overall_SCORE"].min())
)
```

```python
df[["Overall_SCORE", "Overall_Score_Normalized"]].head()
```

```python
df["Performance_Category"] = pd.cut(
    df["Overall_SCORE"],
    bins=[0, 40, 60, 80, 100],
    labels=["Low", "Moderate", "Good", "Excellent"]
)
```

```python
df[["Name", "Overall_SCORE", "Performance_Category"]].head(10)
```

```python
df["Performance_Category"].value_counts()
```

```python
df["Rank_Category"] = pd.cut(
    df["Rank"],
    bins=[0, 10, 50, 100, 500, float("inf")],
    labels=[
        "Top 10",
        "Top 50",
        "Top 100",
        "Top 500",
        "Below 500"
    ]
)
```

```python
df[["Name", "Rank", "Rank_Category"]].head(10)
```

```python
for col in ["the_score", "arwu_score"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    
    df[col + "_Normalized"] = (
        (df[col] - df[col].min()) /
        (df[col].max() - df[col].min())
    )
```

```python
df[
    [
        "the_score",
        "the_score_Normalized",
        "arwu_score",
        "arwu_score_Normalized"
    ]
].head()
```

```python
df.to_csv("Transformed_University_Dataset.csv", index=False)
print("Dataset cleaned successfully")
```

```python
# ==============================
# KPI CALCULATIONS
# ==============================

# 1. Total number of universities
total_universities = df["Name"].nunique()

# 2. Average overall score
average_score = df["Overall_SCORE"].mean()

# 3. Highest overall score
highest_score = df["Overall_SCORE"].max()

# 4. Average QS rank
average_rank = df["Rank"].mean()

# 5. Average international students percentage
average_international_students = df["International_Students_Percentage"].mean()

# 6. Average university age
average_university_age = df["university_age"].mean()

# 7. Number of Top 10 universities
top_10_count = (df["Rank"] <= 10).sum()

# 8. Universities ranked in all 3 ranking systems
ranked_all_3 = df["ranked_in_all_3"].sum()


# ==============================
# CREATE KPI TABLE
# ==============================

kpis = {
    "Total Universities": total_universities,
    "Average Overall Score": round(average_score, 2),
    "Highest Overall Score": round(highest_score, 2),
    "Average QS Rank": round(average_rank, 2),
    "Average International Students %": round(average_international_students, 2),
    "Average University Age": round(average_university_age, 2),
    "Top 10 Universities": top_10_count,
    "Universities Ranked in All 3": ranked_all_3
}

kpi_table = pd.DataFrame(
    list(kpis.items()),
    columns=["KPI", "Value"]
)

kpi_table
```
