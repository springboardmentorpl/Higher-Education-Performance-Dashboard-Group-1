```python
import pandas as pd
```

```python
raw_df = pd.read_csv("university_raw_data.csv")
```

```python
clean_df = raw_df.copy()
```

```python
clean_df = clean_df.replace(r"^\s*\*$", pd.NA, regex=True)
missing_values = ["N/A", "NA", "n/a", "na", "-"]
clean_df = clean_df.replace(missing_values, pd.NA)
```

```python
text_cols = ["University Name", "Country"]
for col in text_cols:
    clean_df[col] = (clean_df[col].astype("string").str.strip().str.replace(r"\s+", " ", regex=True))
clean_df["Country"] = clean_df["Country"].fillna("Unknown")
```

```python
rank_score_cols = ["THE Rank","QS Rank","THE Overall Score","QS Overall Score"]
for col in rank_score_cols:clean_df[col] = (clean_df[col].astype("string").str.strip())
```

```python
performance_cols = [
    "Academic Reputation",
    "Employer Reputation",
    "Teaching Score",
    "Research Score",
    "THE Citation Score",
    "QS Citation Score",
    "Industry Income",
    "Faculty Student Score",
    "International Faculty Score",
    "International Students Score",
    "International Research Network",
    "International Outlook",
    "Employment Outcomes",
    "Sustainability Score"
]
```

```python
numeric_cols = [
    col for col in clean_df.columns
    if col not in text_cols + rank_score_cols
]
for col in numeric_cols:
    clean_df[col] = pd.to_numeric(clean_df[col],errors="coerce")
```

```python
clean_df["THE Rank"] = clean_df["THE Rank"].replace("Reporter",pd.NA)
```

```python
clean_df = clean_df.dropna(subset=performance_cols,how="all")
```

```python
for col in numeric_cols:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())
```

```python
clean_df = clean_df.drop_duplicates()
clean_df = clean_df.reset_index(drop=True)
```

```python
clean_df.to_csv(
    "university_cleaned_data.csv",
    index=False
)
```

```python
print("Original Shape :", raw_df.shape)
print("Cleaned Shape  :", clean_df.shape)
print("\nMissing values after cleaning:")
print(clean_df.isnull().sum())
print("\nRank and Overall Score columns:")
print(clean_df[rank_score_cols].head(20))
```
