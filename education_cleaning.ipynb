```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("university_raw_data.csv")

print("Dataset shape:", df.shape)
```

```python
print("Columns:")
print(df.columns.tolist())
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print("Columns with missing values:")
print(missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

print("Missing value percentage:")
print(missing_percentage.sort_values(ascending=False))
```

```python
columns_to_check = [
    "rank_prefix",
    "score",
    "city",
    "Industry",
    "Overall",
    "Research Environment",
    "Teaching",
    "International Outlook",
    "Research Quality",
    "international_students",
    "faculty_count"
]

for column in columns_to_check:
    print(f"\n--- {column} ---")
    print(df[df[column].isna()][
        ["university", "University_Standard", "country", column]
    ].head(5))
```

```python
print(df.dtypes)
```

```python
print("QS score values:")
print(df["score"].describe())

print("\nExamples of QS scores:")
print(df[["university", "score"]].head(20))
```

```python
print("rank_prefix values:")
print(df["rank_prefix"].value_counts(dropna=False))
```

```python
df = df.drop(columns=["rank_prefix"])

print("Removed rank_prefix column.")
print("New dataset shape:", df.shape)
```

```python
print("Remaining columns:")
print(df.columns.tolist())
```

```python
numeric_columns = [
    "score",
    "Overall",
    "Teaching",
    "Research Environment",
    "Research Quality",
    "Industry",
    "International Outlook"
]

print(df[numeric_columns].describe())
```

```python
print("International students examples:")
print(df["international_students"].dropna().head(20).to_list())

print("\nFaculty count examples:")
print(df["faculty_count"].dropna().head(20).to_list())
```

```python
df["international_students"] = (
    df["international_students"]
    .str.replace(",", "", regex=False)
)

df["international_students"] = pd.to_numeric(
    df["international_students"],
    errors="coerce"
)

print(df["international_students"].dtype)
print(df["international_students"].head(10))
```

```python
df["faculty_count"] = (
    df["faculty_count"]
    .str.replace(",", "", regex=False)
)

df["faculty_count"] = pd.to_numeric(
    df["faculty_count"],
    errors="coerce"
)

print(df["faculty_count"].dtype)
print(df["faculty_count"].head(10))
```

```python
print(df[[
    "international_students",
    "faculty_count"
]].dtypes)
```

```python
errors="coerce"
```

```python
print("Missing international_students:",
      df["international_students"].isna().sum())

print("Missing faculty_count:",
      df["faculty_count"].isna().sum())
```

```python
df["city"] = df["city"].fillna("Unknown")

print("Missing city values:",
      df["city"].isna().sum())
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print("Remaining missing values:")
print(missing)
```

```python
text_columns = df.select_dtypes(include="object").columns

empty_values = {}

for column in text_columns:
    count = (df[column].astype(str).str.strip() == "").sum()
    if count > 0:
        empty_values[column] = count

print("Empty string values:")
print(empty_values)
```

```python
print("Final cleaned dataset shape:", df.shape)
```

```python
duplicate_universities = df.duplicated(
    subset=["University_Standard", "country"]
).sum()

print("Duplicate university-country combinations:",
      duplicate_universities)
```

```python
df.to_csv(
    "university_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
